# Examples, exercises and experiments
## Exercises

```{exercise} Verify the normalisation
:label: ex_pib_norm

Show algebraically that $\psi_n(x) = \sqrt{2/a}\,\sin(n\pi x/a)$ is normalised, i.e., that $\int_0^a |\psi_n|^2\,dx = 1$. Use the identity $\sin^2\theta = \frac{1}{2}(1-\cos 2\theta)$.
```

```{exercise} Energy level spacing 
:label: ex_pib_spacing

Show that the spacing between adjacent energy levels is $E_{n+1} - E_n = (2n+1)E_1$.

1. Is the spacing constant (equally spaced levels)?
2. What happens to the *fractional* spacing $(E_{n+1}-E_n)/E_n$ as $n \to \infty$?
3. Relate your answer to part (2) to the correspondence principle.
```

```{exercise} Zero-point energy and box size  
:label: ex_pib_zpe

An electron ($m_e = 9.109\times10^{-31}\;\mathrm{kg}$) is confined to a 1D box.

1. Calculate the zero-point energy $E_1$ for a box of length $a = 1\;\mathrm{nm}$ (roughly the size of a small molecule). Express your answer in eV ($1\;\mathrm{eV} = 1.602\times10^{-19}\;\mathrm{J}$).

2. Calculate $E_1$ for a box of length $a = 1\;\mathrm{m}$ (macroscopic). Why is the zero-point energy negligible on a macroscopic scale?

3. For what box length $a$ does $E_1 = k_BT$ at room temperature ($T = 300\;\mathrm{K}$, $k_B = 1.381\times10^{-23}\;\mathrm{J/K}$)?
```

```{exercise} Probability in a region  
:label: ex_pib_prob

For a particle in a 1D box of length $a$:

1. Find the probability of locating the particle in the middle third of the box ($a/3 \leq x \leq 2a/3$) for $n = 1$ and $n = 2$.

2. For the $n=1$ state, is the particle more likely to be found in the middle third or the outer thirds? Does this make sense given the shape of $|\psi_1|^2$?

3. For large $n$, what value would you expect the probability of finding the particle in any third of the box to approach? Explain.
```

```{exercise} Expectation values   
:label: ex_pib_exp

For the $n$th state of the particle in a 1D box:

1. Verify analytically that $\langle p \rangle_n = 0$ using $\hat{p} = -i\hbar\,d/dx$.

2. Compute $\langle p^2 \rangle_n$ and show that $\langle p^2 \rangle_n / (2m) = E_n$. Interpret this result physically.

3. Show that $\Delta x\,\Delta p > \hbar/2$ for all $n \geq 1$.
```

```{exercise} Wavefunction superposition   
:label: ex_pib_super

A particle in a box of length $a$ is prepared at $t=0$ in the state:

$$\Phi(x,0) = \frac{1}{\sqrt{2}}\psi_1(x) + \frac{1}{\sqrt{2}}\psi_3(x)$$

1. Verify that $\Phi(x,0)$ is normalised.
2. Write down the full time-dependent wavefunction $\Phi(x,t)$.
3. Compute $\langle E \rangle$ (the average energy). Is this one of the allowed energies $E_n$?
4. Compute the probability density $|\Phi(x,t)|^2$ and show that it oscillates in time. Find the oscillation period $T$.
5. Write a Python script to animate $|\Phi(x,t)|^2$ over one period.
```

```{exercise} $\beta$-carotene absorption   
:label: ex_pib_carotene

Use the 1D particle-in-a-box model to estimate the lowest-energy electronic absorption of $\beta$-carotene.

Given:
- Conjugated chain length: $L = 2.4\;\mathrm{nm}$
- Number of $\pi$ electrons: 22 (filling levels $n=1$ through $n=11$)
- $m_e = 9.109\times10^{-31}\;\mathrm{kg}$, $h = 6.626\times10^{-34}\;\mathrm{J\cdot s}$

1. Calculate $\Delta E$ for the $n=11 \to n=12$ transition.
2. Calculate the corresponding wavelength $\lambda = hc/\Delta E$.
3. The experimental absorption maximum is $\approx 450\;\mathrm{nm}$. Calculate the percentage error of your model and suggest two physical reasons why the model is imperfect.
```

```{exercise} Quantum dot engineering   
:label: ex_pib_qdot

A CdSe quantum dot can be modelled as a 3D cubic PIB with side length $d$ and an effective electron mass of $m^* = 0.13\,m_e$.

1. Derive the expression for the ground-state-to-first-excited-state transition energy $\Delta E = 3h^2/(8m^*d^2)$.
2. Calculate $d$ such that the dot emits green light ($\lambda = 530\;\mathrm{nm}$).
3. If a second dot must emit red light ($\lambda = 650\;\mathrm{nm}$), what should its diameter be?
4. Write a Python script that plots the emission wavelength vs. dot diameter for $d$ in the range $1$–$10\;\mathrm{nm}$.
```

```{exercise} 2D square box and degeneracy    
:label: ex_pib_2d

Consider a particle confined to a 2D square box of side $a$.

1. Write down $\psi_{n_x,n_y}(x,y)$ and $E_{n_x,n_y}$ following the separation-of-variables approach.
2. List all states with energies up to and including $13E_1$ (where $E_1 = h^2/(8ma^2)$), and their degeneracies.
3. Write a Python script that creates a colour-map plot of $|\psi_{n_x,n_y}|^2$ for several $(n_x,n_y)$ combinations. Label each plot with its quantum numbers and energy.
4. Now consider a *rectangular* box with sides $a$ and $2a$. Do the $(2,1)$ and $(1,2)$ states remain degenerate? Explain physically.
```

```{exercise} Numerical integration verification  
:label: ex_pib_numerical

Using Python:

1. Numerically verify the orthonormality relation $\int_0^a \psi_m^*\psi_n\,dx = \delta_{mn}$ for all pairs $(m,n)$ with $m,n \in \{1,2,3,4,5\}$. Display your results as a $5\times5$ matrix (which should be close to the identity matrix).
2. Numerically compute $\langle x \rangle_n$, $\langle x^2 \rangle_n$, and $\Delta x_n$ for $n = 1$ to $10$ and compare with the analytical results. Plot $\Delta x_n$ vs. $n$.
3. Numerically verify $\Delta x_n \cdot \Delta p_n \geq \hbar/2$ for $n = 1$ to $10$. Plot the uncertainty product vs. $n$.
```
