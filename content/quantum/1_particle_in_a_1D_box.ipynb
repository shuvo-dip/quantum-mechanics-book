# Examples, exercises and experiments
## Exercises

```{exercise} Verify the normalisation
:label: ex_pib_norm

Show algebraically that $\psi_n(x) = \sqrt{2/a}\,\sin(n\pi x/a)$ is normalised, i.e., that $\int_0^a |\psi_n|^2\,dx = 1$. Use the identity $\sin^2\theta = \frac{1}{2}(1-\cos 2\theta)$.
```

```{exercise} Energy level spacing 
:label: ex_pib_spacing

Show that the spacing between adjacent energy levels is $E_{n+1} - E_n = (2n+1)E_1$.

1. Is the spacing constant (equally spaced levels)?
2. What happens to the *fractional* spacing $(E_{n+1}-E_n)/E_n$ as $n \to \infty$?
3. Relate your answer to part (2) to the correspondence principle.
```

```{exercise} Zero-point energy and box size  
:label: ex_pib_zpe

An electron ($m_e = 9.109\times10^{-31}\;\mathrm{kg}$) is confined to a 1D box.

1. Calculate the zero-point energy $E_1$ for a box of length $a = 1\;\mathrm{nm}$ (roughly the size of a small molecule). Express your answer in eV ($1\;\mathrm{eV} = 1.602\times10^{-19}\;\mathrm{J}$).

2. Calculate $E_1$ for a box of length $a = 1\;\mathrm{m}$ (macroscopic). Why is the zero-point energy negligible on a macroscopic scale?

3. For what box length $a$ does $E_1 = k_BT$ at room temperature ($T = 300\;\mathrm{K}$, $k_B = 1.381\times10^{-23}\;\mathrm{J/K}$)?
```

```{exercise} Probability in a region  
:label: ex_pib_prob

For a particle in a 1D box of length $a$:

1. Find the probability of locating the particle in the middle third of the box ($a/3 \leq x \leq 2a/3$) for $n = 1$ and $n = 2$.

2. For the $n=1$ state, is the particle more likely to be found in the middle third or the outer thirds? Does this make sense given the shape of $|\psi_1|^2$?

3. For large $n$, what value would you expect the probability of finding the particle in any third of the box to approach? Explain.
```

```{exercise} Expectation values   
:label: ex_pib_exp

For the $n$th state of the particle in a 1D box:

1. Verify analytically that $\langle p \rangle_n = 0$ using $\hat{p} = -i\hbar\,d/dx$.

2. Compute $\langle p^2 \rangle_n$ and show that $\langle p^2 \rangle_n / (2m) = E_n$. Interpret this result physically.

3. Show that $\Delta x\,\Delta p > \hbar/2$ for all $n \geq 1$.
```
